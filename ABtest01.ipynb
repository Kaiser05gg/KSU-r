```r
# R入門：ベクトル/関数/記述統計量
x <- c(2, 4, 6, 8, 10)                    # ベクトル
y <- x * 2 + 1                            # ベクトル演算

# 出力
print(x)
print(y)

# 記述統計量の例
mean(y)                                   # 平均
var(y)                                    # 分散
sd(y)                                     # 標準偏差

# 見やすくする出力方法：cat()             # concatenate
cat("mean=", mean(y), " sd=", sd(y), "\n")
```

```r
# サイコロ：期待値と分散
x <- 1:6
p <- rep(1/6, 6)                         # 確率
mu <- sum(x * p)                         # 期待値（*：ベクトルに使えば、内積（成分ごとの積和）の計算をする）
v  <- sum((x - mu)^2 * p)                # 分散（^2：成分ごとに2乗する）

# 出力
cat("E[X]=", mu, " Var[X]=", v, "\n")    # 出力
```

```r
# サマージャンボ宝くじの期待値と分散
prize <- c(5e8, 1e8, 1e5, 1e7, 1e6, 1e5, 3000, 300)   # 1等, 前後賞, 組違い, 2等, 3等, 4等, 5等, 6等
count <- c(1,   2,   99,  2,   30,  600, 1e5,  1e6)   # 各「1ユニットあたり」の当せん本数
N <- 1e7                                              # 1ユニット = 1,000万枚
p <- count / N                                        # 1枚あたり当せん確率
mu <- sum(prize * p)                                  # 期待値

# 出力
cat("E[当選金額]=",mu,"\n")
```

```r
# 一様分布の可視化
x <- seq(0, 1, length=200)                      # グリッド点をxとして
y <- dunif(x, min=0, max=1)                     # 密度関数（理論上）をグリッド上に、yとして

# 可視化
hist(runif(50, 0, 1))

hist(runif(50, 0, 1), prob=TRUE, xlab="x")      # サンプル 50 個のヒストグラム, "prob=True"で相対度数にして出力
lines(x, y, col="red", lwd=2)                   # 理論密度線を重ねる（※ヒストグラムの直後のlinesで自動的に重ね付け）

hist(runif(5000, 0, 1), prob=TRUE, breaks=30, main="U(0,1) & density(red)", xlab="x") # サンプル 5000 個のヒストグラム, "prob=True"で相対度数にして出力
lines(x, y, col="red", lwd=2)                   # 理論密度線を重ねる（※ヒストグラムの直後のlinesで自動的に重ね付け）

# 何度か出力し直してみてください
```

```r
# 手入力で次のことをやってみましょう
# １．大きさ100,000の一様分布からの乱数を可視化してください（hist()を使うこと）
# ２．タイトルをつけて、ビンの数を50にして、相対度数にして、x軸を"x"と名付けて、もう一度可視化してください


```

```r
# 二項分布の可視化
n <- 50                  # ベルヌーイ試行を50回
p <- 0.5                 # 成功確率0.5
x <- rbinom(5000, n, p)  # 二項分布から値を5000個ドローして格納

# 出力
cat("mean(x)=", mean(x)) # 平均

# ヒストグラム（ビン、タイトル、軸ラベルを指定）
hist(x, breaks=seq(-0.5, n+0.5, 1), main="B(n=50, p=0.4)", xlab="k")
```

```r
# 二項分布に正規分布を重ねて表示（中心極限定理）
mu <- n * p               # 期待値（理論的上の期待値）
s <- sqrt(n * p * (1-p))  # 母集団標準偏差（理論上の標準偏差）

hist(x, prob=TRUE, breaks=seq(-0.5, n+0.5, 1), main="Binomial(n=50,p=0.4)", xlab="k") # 相対度数に変換してヒストグラム
curve(dnorm(x, mu, s), add=TRUE, col="red") # 正規分布の密度関数を赤色で重ね付け

### 二項分布が正規分布になっていることを確認しましょう（むしろ、これを正規分布の定義と見なす流儀もある）
```
