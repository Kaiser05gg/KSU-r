### 関数

これは数学の言葉です．もうこの時点で「この授業やめよかな」と思うかもしれませんが，やめないでください．大丈夫です．名前は関数ですし，数学で出てきた関数と同じ意味ですが，おそらくこっちのほうがとっつきやすいはずです．

まず以下の「関数」を定義ましょう．
$$
y = 2x + 5
$$
一次関数です．右肩上がりで，傾きが2，切片が5です．この意味は，「$x$はなんか知らんけど，2かけて5足したらそれが$y$やで」です．つまり計算方法を記述しているのです．あくまで計算方法ですから，$x$として何らかの具体的な数値を指定しないと，いつまでたっても$y$が具体的に決まりません．$x=1$とすれば$y=7$になりますが，そうしない限り永遠に$y$は謎のままです．これが関数です．Rでこの関数を定義しましょう．


```r
myfunc <- function(x){
  return (2*x+5)
}
```

何も出力されなくて正解です．これは関数を定義しただけで，**使っていない**のです．「`x`を誰かから指定してもらってください．その場合は，`2*x+5`の値を返してあげますから」という意味です．この関数に`myfunc`という名前を付けました．名前自体は何でもいいんです．`ogoharakimoi`でも，`aho`でも，`INeedCredit`でも，好きにしてください．そうすれば，**$x$に何を指定しようと，とにかく$2x+5$を計算する**関数をいつでも利用可能です．

> 一つ注意です．上の関数を定義した3行のプログラムのうち1行目と2行目に`x`が出てきます．この2つの`x`は，さらに上の節で使用した`x <- seq( 0, 360, by=60 )`とは全く別物です．この関数定義部分で使用する文字は，この関数定義部分で一貫していればいいので，`a`でも`p`でも`w`でもなんでもいいです．

```r
myfunc(1)
```

```r
myfunc(-10)
```

```r
myfunc(75892783459072)
```

もちろん，`x`に配列を指定してもいいんです．

```r
myfunc(y)
```
